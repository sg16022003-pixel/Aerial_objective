```python
import zipfile
import os

# Define the target extraction directory
extraction_base_dir = '/content/extracted_dataset'
os.makedirs(extraction_base_dir, exist_ok=True)

# Paths to the zip files (re-using variables from fkjg-VImvxVd)
# Using the correct zip file names identified earlier
train_zip_path = '/content/train-20260927T063937Z-1-001.zip'
val_zip_path =  '/content/valid-20260927T063922Z-1-001.zip'
test_zip_path = '/content/test-20260927T063951Z-1-001.zip'

def extract_zip(zip_path, extract_to_dir):
    if os.path.exists(zip_path):
        print(f"Extracting {zip_path} to {extract_to_dir}...")
        try:
            with zipfile.ZipFile(zip_path, 'r') as zip_ref:
                zip_ref.extractall(extract_to_dir)
            print("Extraction complete.")
        except zipfile.BadZipFile:
            print(f"Error: {zip_path} is not a valid zip file. Please check its integrity or re-upload.")
        except Exception as e:
            print(f"An unexpected error occurred while extracting {zip_path}: {e}")
    else:
        print(f"Warning: {zip_path} not found. Please ensure it is uploaded to /content/.")

# Extract each zip file into its own subdirectory within extracted_dataset
extract_zip(train_zip_path, os.path.join(extraction_base_dir, 'train_root'))
extract_zip(val_zip_path, os.path.join(extraction_base_dir, 'valid_root'))
extract_zip(test_zip_path, os.path.join(extraction_base_dir, 'test_root'))

print("All specified zip files have been processed for extraction.")
print("Please verify that the files were extracted correctly by checking the '/content/extracted_dataset' directory.")

# --- Diagnostic: Check contents of extracted directories ---
print("\n--- Verifying extracted contents ---")
for split_type in ['train', 'valid', 'test']:
    extracted_split_base = os.path.join(extraction_base_dir, f'{split_type}_root', split_type)
    bird_path = os.path.join(extracted_split_base, 'bird')
    drone_path = os.path.join(extracted_split_base, 'drone')

    print(f"Checking {split_type} extracted data:")
    if os.path.exists(extracted_split_base):
        print(f"  Base extracted path '{extracted_split_base}' exists.")
        if os.path.exists(bird_path) and os.path.isdir(bird_path):
            bird_files = [f for f in os.listdir(bird_path) if os.path.isfile(os.path.join(bird_path, f))]
            print(f"    '{bird_path}' contains {len(bird_files)} files.")
            if len(bird_files) > 0: print(f"      Sample files: {bird_files[:5]}") # Print first 5 files
        else:
            print(f"    '{bird_path}' (bird class directory) not found.")

        if os.path.exists(drone_path) and os.path.isdir(drone_path):
            drone_files = [f for f in os.listdir(drone_path) if os.path.isfile(os.path.join(drone_path, f))]
            print(f"    '{drone_path}' contains {len(drone_files)} files.")
            if len(drone_files) > 0: print(f"      Sample files: {drone_files[:5]}")
        else:
            print(f"    '{drone_path}' (drone class directory) not found.")
    else:
        print(f"  Base extracted path '{extracted_split_base}' not found.")
```

```python
import os
import shutil

# Define paths
extraction_base_dir = '/content/extracted_dataset'
final_base_dir = '/content/final_dataset'

# Create the final base directory if it doesn't exist
os.makedirs(final_base_dir, exist_ok=True)

print(f"Organizing data from '{extraction_base_dir}' to '{final_base_dir}'...")

for split_type in ['train', 'valid', 'test']:
    intermediate_source_dir = os.path.join(extraction_base_dir, f'{split_type}_root')
    destination_split_dir = os.path.join(final_base_dir, split_type)

    # Clean up the destination directory for this split
    if os.path.exists(destination_split_dir):
        shutil.rmtree(destination_split_dir)
        print(f"Cleaned existing directory: {destination_split_dir}")
    os.makedirs(destination_split_dir, exist_ok=True)
    print(f"Created directory: {destination_split_dir}")

    found_class_data_root = None
    # Walk through the directory tree to find the correct parent of 'bird' and 'drone'
    for root, dirs, files in os.walk(intermediate_source_dir):
        # Check if this 'root' directory contains both 'bird' and 'drone' subdirectories
        if 'bird' in dirs and 'drone' in dirs:
            # Additionally, ensure that the current 'root' corresponds to the split_type structure.
            # For example, for 'train', we expect /content/extracted_dataset/train_root/train
            # This makes sure we are not picking up class folders from a deeper, incorrect level if they exist.
            if os.path.basename(root) == split_type or os.path.basename(os.path.dirname(root)) == split_type: # Added robustness
                found_class_data_root = root
                break  # Found the correct directory containing both class folders

    if found_class_data_root:
        print(f"Copying 'bird' and 'drone' data from {found_class_data_root} to {destination_split_dir} for {split_type} split...")
        # Copy only the 'bird' and 'drone' subdirectories from the identified root
        shutil.copytree(os.path.join(found_class_data_root, 'bird'), os.path.join(destination_split_dir, 'bird'), dirs_exist_ok=True)
        shutil.copytree(os.path.join(found_class_data_root, 'drone'), os.path.join(destination_split_dir, 'drone'), dirs_exist_ok=True)
        print(f"Finished copying for {split_type}.")
    else:
        print(f"Warning: Could not find 'bird' and 'drone' class directories under {intermediate_source_dir} for {split_type}. No data copied.")

print("Data organization complete.")
print("\n--- Verifying final dataset structure ---")
for split in ['train', 'valid', 'test']:
    path = os.path.join(final_base_dir, split)
    if os.path.exists(path):
        print(f"Contents of {path}:", os.listdir(path))
        bird_path = os.path.join(path, 'bird')
        drone_path = os.path.join(path, 'drone')
        if os.path.exists(bird_path):
            print(f"  '{split}/bird' has {len(os.listdir(bird_path))} items.")
        else:
            print(f"  Warning: '{split}/bird' does not exist.")
        if os.path.exists(drone_path):
            print(f"  '{split}/drone' has {len(os.listdir(drone_path))} items.")
        else:
            print(f"  Warning: '{split}/drone' does not exist.")
    else:
        print(f"Warning: Directory '{path}' does not exist.")
```

```python
import os
import shutil

BASE_DIR = '/content/final_dataset'

for split in ['train', 'valid', 'test']:
    # Path to the nested subfolders
    nested_path = os.path.join(BASE_DIR, split, split, 'images')
    target_path = os.path.join(BASE_DIR, split)

    if os.path.exists(nested_path):
        print(f"Flattening structure for {split}...")
        for item in os.listdir(nested_path):
            src = os.path.join(nested_path, item)
            dst = os.path.join(target_path, item)
            shutil.move(src, dst)

        # Clean up the old nested folder
        shutil.rmtree(os.path.join(BASE_DIR, split, split))
        print(f"Successfully organized {split}!")
    else:
        print(f"Skipping flattening for {split}: Nested path '{nested_path}' not found. This might indicate an issue with previous extraction or organization.")

print("\nUpdated structure verification:")
for split in ['train', 'valid', 'test']:
    path = os.path.join(BASE_DIR, split)
    if os.path.exists(path):
        print(f"Contents of {path}:", os.listdir(path))
    else:
        print(f"Warning: Directory '{path}' does not exist.")
```

```python
import os
import tensorflow as tf

BASE_DIR = '/content/final_dataset'
TRAIN_DIR = os.path.join(BASE_DIR, 'train')
VALID_DIR = os.path.join(BASE_DIR, 'valid')
TEST_DIR = os.path.join(BASE_DIR, 'test')

# Load and verify
for name, path in [('Train', TRAIN_DIR), ('Validation', VALID_DIR), ('Test', TEST_DIR)]:
    print(f"\n--- Verifying {name} set at: {path} ---")
    if not os.path.exists(path):
        print(f"Error: Directory '{path}' does not exist. Skipping {name} set.")
        continue
    if not os.path.isdir(path):
        print(f"Error: '{path}' is not a directory. Skipping {name} set.")
        continue

    dir_contents = os.listdir(path)
    print(f"Contents of '{path}': {dir_contents}")

    # Check for expected class subdirectories
    expected_classes = ['bird', 'drone'] # Assuming these are the class names based on problem description
    found_classes_with_files = []

    # Flag to detect if images are directly in the split directory
    image_files_in_root = [f for f in dir_contents if os.path.isfile(os.path.join(path, f)) and f.lower().endswith(('.png', '.jpg', '.jpeg', '.gif', '.bmp'))]
    if image_files_in_root:
        print(f"Warning: Found {len(image_files_in_root)} image files directly in '{path}'.\n         tf.keras.utils.image_dataset_from_directory expects class subdirectories for label inference. This will likely cause 'class_names=[]'.")

    for class_name_expected in expected_classes:
        class_path = os.path.join(path, class_name_expected)
        if os.path.isdir(class_path):
            class_files = [f for f in os.listdir(class_path) if os.path.isfile(os.path.join(class_path, f)) and f.lower().endswith(('.png', '.jpg', '.jpeg', '.gif', '.bmp'))]
            print(f"  Class directory '{class_name_expected}' exists with {len(class_files)} recognized image files.")
            if len(class_files) > 0:
                found_classes_with_files.append(class_name_expected)
        else:
            print(f"  Class directory '{class_name_expected}' does NOT exist in '{path}'.")

    if len(found_classes_with_files) < 2 and not image_files_in_root:
        print(f"Error: Not enough class directories with recognized image files found in '{path}' for binary classification. Found: {found_classes_with_files}. Skipping {name} set.")
        continue
    elif len(found_classes_with_files) == 0 and not image_files_in_root:
        print(f"Error: No class directories with recognized image files found in '{path}'. Cannot infer classes. Skipping {name} set.")
        continue

    ds = None # Initialize ds before the try block
    try:
        ds = tf.keras.utils.image_dataset_from_directory(
            path,
            image_size=(224, 224),
            batch_size=32,
            label_mode='binary',
            verbose=False # Suppress default output, we're adding our own diagnostics
        )
        print(f"--- {name} set loaded successfully! Classes: {ds.class_names}, Found {ds.cardinality().numpy() * ds.batch_size} images ---")
    except ValueError as e:
        print(f"Error loading {name} set with image_dataset_from_directory: {e}")
        # Explicitly print inferred class_names if the error is related to it
        if ds is not None: # Check if ds was assigned before trying to access its attributes
            print(f"  Inferred classes by image_dataset_from_directory: {getattr(ds, 'class_names', 'Error during inference')}")
        else:
            print("  Dataset object 'ds' was not initialized due to early error or no classes found.")
    except Exception as e:
        print(f"An unexpected error occurred while loading {name} set: {e}")
```

```python
import os
import tensorflow as tf

BASE_DIR = '/content/final_dataset'

TRAIN_DIR = os.path.join(BASE_DIR, 'train')
VALID_DIR = os.path.join(BASE_DIR, 'valid')
TEST_DIR = os.path.join(BASE_DIR, 'test')

IMG_SIZE = (224, 224)
BATCH_SIZE = 32

print("Loading datasets...")

train_ds = tf.keras.utils.image_dataset_from_directory(
    TRAIN_DIR,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode='binary'
)

valid_ds = tf.keras.utils.image_dataset_from_directory(
    VALID_DIR,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode='binary'
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    TEST_DIR,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode='binary',
    shuffle=False
)

print(f"\n✓ Datasets loaded successfully!")
print(f"Classes found: {train_ds.class_names}")
```

```python
AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.cache().prefetch(buffer_size=AUTOTUNE)
valid_ds = valid_ds.cache().prefetch(buffer_size=AUTOTUNE)
```

```python

from tensorflow.keras import layers, models
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint

# Augmentation pipeline
data_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.15),
    layers.RandomZoom(0.15),
])

# Transfer Learning Setup
base_model = MobileNetV2(input_shape=(224, 224, 3), include_top=False, weights='imagenet')
base_model.trainable = False

inputs = layers.Input(shape=(224, 224, 3))
x = data_augmentation(inputs)
x = preprocess_input(x)
x = base_model(x, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.3)(x)
outputs = layers.Dense(1, activation='sigmoid')(x)

model = models.Model(inputs, outputs)

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

callbacks = [
    EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True),
    ModelCheckpoint('best_aerial_model.h5', monitor='val_loss', save_best_only=True)
]

history = model.fit(
    train_ds,
    validation_data=valid_ds,
    epochs=15,
    callbacks=callbacks
)

# Plotting training and validation accuracy and loss
fig, ax = plt.subplots(1, 2, figsize=(16, 6))

# Plot training & validation accuracy values
ax[0].plot(history.history['accuracy'])
ax[0].plot(history.history['val_accuracy'])
ax[0].set_title('Model accuracy')
ax[0].set_ylabel('Accuracy')
ax[0].set_xlabel('Epoch')
ax[0].legend(['Train', 'Validation'], loc='upper left')

# Plot training & validation loss values
ax[1].plot(history.history['loss'])
ax[1].plot(history.history['val_loss'])
ax[1].set_title('Model loss')
ax[1].set_ylabel('Loss')
ax[1].set_xlabel('Epoch')
ax[1].legend(['Train', 'Validation'], loc='upper left')

plt.tight_layout()
plt.show()
```

```python
%%writefile streamlit_app.py
import os
import streamlit as st
import numpy as np
import tensorflow as tf
from PIL import Image

# Explicitly tell TensorFlow to use CPU only by setting environment variable
os.environ['CUDA_VISIBLE_DEVICES'] = '-1'
tf.config.set_visible_devices([], 'GPU')

# Page Configuration
st.set_page_config(
    page_title="Aerial Object Classifier",
    page_icon="🚁",
    layout="centered"
)

st.title("🚁 Aerial Object Classification App")
st.write("Upload an aerial photograph to classify whether it contains a **Bird** or a **Drone**.")

# Cache Model Loading
@st.cache_resource
def load_trained_model():
    return tf.keras.models.load_model('best_aerial_model.keras')

model = load_trained_model()
CLASS_NAMES = ['Bird', 'Drone']

# Image Upload
uploaded_file = st.file_uploader("Choose an aerial image...", type=["jpg", "jpeg", "png"])

if uploaded_file is not None:
    # 1. Loading the image
    image = Image.open(uploaded_file)
    st.image(image, caption="Uploaded Image", use_container_width=True)

    # 2. Resizing to match what the model trained on
    image = image.resize((224, 224))

    # 3. Converting to an array and expanding dimensions
    img_array = np.array(image)
    img_array = np.expand_dims(img_array, axis=0)

    # 4. Apply MobileNetV2 preprocessing
    img_preprocessed = tf.keras.applications.mobilenet_v2.preprocess_input(img_array)

    # 5. Perform Inference
    prediction = model.predict(img_preprocessed)[0][0]

    # Calculate confidence and thresholds
    confidence = float(prediction)

    # 6. Display Predictions with Ambiguity Threshold
    if confidence > 0.75:
        st.success("Prediction: **Drone**")
        st.progress(confidence)
        st.info(f"**Confidence Score:** {confidence * 100:.2f}%")
    elif confidence < 0.25:
        bird_conf = 1.0 - confidence
        st.success("Prediction: **Bird**")
        st.progress(bird_conf)
        st.info(f"**Confidence Score:** {bird_conf * 100:.2f}%")
    else:
        max_conf = max(confidence, 1.0 - confidence)
        st.warning("Prediction: **Uncertain / Ambiguous Object**")
        st.progress(max_conf)
        st.info(f"**Confidence Score:** {max_conf * 100:.2f}% (Too low to classify definitively)")
```
